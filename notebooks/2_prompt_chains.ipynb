```python
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
out=StrOutputParser()
```

```python
####Myself chain node
def transform_case(result:str):
    return result.upper()
```

```python
prompts=ChatPromptTemplate.from_messages([
    {
        "role":"system", "content":"You are a translator translate input in {Language}"},
        {"role":"user", "content":"{Query}"},
    
])

```

```python
## Chains- Runnable


res=llm.invoke(prompts.invoke({"Language":"Hindi", "Query":"I love you?"}))
print(res.content)
```

```python
chains=prompts | llm | out| transform_case
```

```python
res=chains.invoke({"Language":"English", "Query":"I love langchain"})
print(res)
```
